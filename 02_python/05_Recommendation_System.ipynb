# Product Recommendation System

## Objective

The objective of this analysis is to develop a product recommendation
system using customer purchase behavior.

The system identifies products that are commonly purchased by customers
with similar purchasing patterns and recommends relevant products.

## Approach

1. Load the cleaned Lulu Mall sales dataset.
2. Create a customer-product interaction matrix.
3. Calculate similarity between products based on customer purchases.
4. Generate product recommendations.
5. Display recommended products for selected products.

step 1: Load Data

```python
import pandas as pd
import numpy as np
df = pd.read_excel( r"D:\Lulu_Mall_Retail_Analytics recent\Lulu_Mall_Sales_Dataset.xlsx")
df.head()
```

Step 2 Check the required columns

```python
df[["Customer_ID", "Product_Name", "Quantity"]].head()
```

Step 3: Create Customer × Product matrix

```python
customer_product_matrix = df.pivot_table(
    index="Customer_ID",
    columns="Product_Name",
    values="Quantity",
    aggfunc="sum",
    fill_value=0
)
customer_product_matrix.head()
```

Step 4: Calculate product similarity

## Product Similarity

Cosine similarity is used to measure how similar two products are
based on customer purchasing behavior.

Products purchased by similar groups of customers will receive
higher similarity scores.

```python
from sklearn.metrics.pairwise import cosine_similarity

product_similarity = cosine_similarity(customer_product_matrix.T)

product_similarity_df = pd.DataFrame(
    product_similarity,
    index=customer_product_matrix.columns,
    columns=customer_product_matrix.columns
)

product_similarity_df.head()
```

## Recommendation function
STEP 5

```python
def recommend_products(product_name, n=5):
    
    if product_name not in product_similarity_df.index:
        return f"Product '{product_name}' not found in dataset."
    
    similar_products = product_similarity_df[product_name].sort_values(
        ascending=False
    )
    
    recommendations = similar_products.iloc[1:n+1]
    
    return recommendations
```

```python
df["Product_Name"].unique()
```

```python
recommendations = recommend_products("Samsung Galaxy Tab S9", 5)
recommendations
```

# Personalized Recommendation System

The recommendation system supports two recommendation approaches:

1. Product-based recommendations
2. Customer-based personalized recommendations

Product-based recommendations identify products with similar
customer purchasing patterns.

Customer-based recommendations identify products that a customer
has not purchased yet but are frequently purchased by customers
with similar purchasing behavior.

customer-based recommendations

```python
def recommend_for_customer(customer_id, n=5):

    if customer_id not in customer_product_matrix.index:
        return pd.DataFrame(
            columns=["Product_Name", "Score"]
        )

    customer_vector = customer_product_matrix.loc[
        customer_id
    ].values.reshape(1, -1)

    customer_similarity = cosine_similarity(
        customer_vector,
        customer_product_matrix.values
    )[0]

    similarity_scores = pd.Series(
        customer_similarity,
        index=customer_product_matrix.index
    )

    similarity_scores = similarity_scores.drop(
        customer_id,
        errors="ignore"
    )

    similar_customers = similarity_scores.sort_values(
        ascending=False
    )

    purchased_products = set(
        customer_product_matrix.loc[customer_id]
        .loc[lambda x: x > 0]
        .index
    )

    scores = {}

    for similar_customer, similarity in similar_customers.head(20).items():

        products = customer_product_matrix.loc[
            similar_customer
        ]

        for product, quantity in products.items():

            if quantity > 0 and product not in purchased_products:

                scores[product] = (
                    scores.get(product, 0)
                    + similarity * quantity
                )

    recommendations = (
        pd.Series(scores)
        .sort_values(ascending=False)
        .head(n)
        .reset_index()
    )

    recommendations.columns = [
        "Product_Name",
        "Score"
    ]

    return recommendations
```
