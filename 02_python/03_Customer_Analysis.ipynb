**Customer Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("output/Lulu_Mall_Sales_Cleaned.csv")
```

Customer summary

```python
customer_summary = (
    df.groupby("Customer_ID")
      .agg(
          Total_Orders=("Order_ID", "nunique"),
          Total_Revenue=("Revenue", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Last_Purchase=("Order_Date", "max")
      )
      .reset_index()
)
```

```python
customer_summary.head()
```

```python
customer_summary["Average_Order_Value"] = (
    customer_summary["Total_Revenue"] /
    customer_summary["Total_Orders"]
)
```

**Customer segmentation**

```python
def segment_customer(row):

    if row["Total_Revenue"] >= 50000:
        return "High Value"
    elif row["Total_Revenue"] >= 25000:
        return "Regular"
    elif row["Total_Revenue"] >= 10000:
        return "Occasional"
    else:
        return "Low Value"

customer_summary["Customer_Segment"] = (
    customer_summary.apply(segment_customer, axis=1)
)
```
